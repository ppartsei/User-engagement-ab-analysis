# Разработка A/B-тестирования и анализ результатов

Вы работаете в компании развлекательного приложения с «бесконечной» лентой коротких видео. Основная цель продукта — удерживать внимание пользователя.

Монетизация приложения строится на двух моделях доступа к видео:

* Рекламная модель — бесплатный доступ к контенту, доход приносят рекламные ролики, встроенные в ленту.
* Подписочная модель — ежемесячная оплата, которая полностью отключает рекламу и делает просмотр беспрерывным.
Главный двигатель вашего приложения — это алгоритм рекомендаций.


>Рекомендательная система — это алгоритм, который из миллионов доступных вариантов выбирает те, которые с наибольшей вероятностью понравятся конкретному пользователю.
>Если алгоритм в приложении для коротких видео будет работать плохо (то есть будет отбирать неинтересный контент), пользователь быстро его закроет. Если хорошо — это создаст для пользователя стимул остаться в приложении дольше, что напрямую влияет на выручку компании.


До недавнего времени лента формировалась на основе простых эвристик, по схеме «популярное за день + новинки». Сейчас всё меняется: команда Data Science разработала новый алгоритм ранжирования контента.

В отличие от текущей системы, модель МО учитывает персональные интересы пользователя: его прошлые отметки «Нравится» и длительность просмотра. Модель оптимизирована на предсказание вероятности глубокого вовлечения (просмотр более 80% видео). Таким образом, более релевантный контент должен увеличивать время пользователей в приложении. Однако есть два риска, которые могут негативно повлиять на доходы:

1. Пользователи без подписки, максимально поглощённые релевантным контентом, могут начать негативно реагировать на рекламные вставки.
2. Ценность платной подписки «без рекламы» снизится, если контент станет слишком поглощающим.

````python
#  Агрегация данных по дням
# Считаем уникальных пользователей (DAU) и зарегистрированных пользователей за каждый день
daily_stats = (
    sessions_history.groupby("session_date")
    .agg(
        total_users=("user_id", "nunique"),
        registered_users=(
            "user_id",
            lambda x: sessions_history.loc[x.index, "user_id"][
                sessions_history.loc[x.index, "registration_flag"] == 1
            ].nunique(),
        ),
    )
    .reset_index()
)

# Сортируем по дате, чтобы графики строились последовательно
daily_stats = daily_stats.sort_values("session_date")

# Рассчитываем долю зарегистрированных пользователей от общего числа
daily_stats["registration_share"] = (
    daily_stats["registered_users"] / daily_stats["total_users"]
)


#  График 1: Общее число пользователей и число зарегистрированных пользователей
plt.figure(figsize=(12, 6))
plt.plot(
    daily_stats["session_date"],
    daily_stats["total_users"],
    label="Все пользователи (DAU)",
    color="blue",
    linewidth=2,
)
plt.plot(
    daily_stats["session_date"],
    daily_stats["registered_users"],
    label="Зарегистрированные пользователи",
    color="green",
    linewidth=2,
)

plt.title(
    "Динамика общего числа пользователей и регистраций по дням", fontsize=14
)
plt.xlabel("Дата сессии", fontsize=12)
plt.ylabel("Количество уникальных пользователей", fontsize=12)
plt.xticks(rotation=45)  # Наклоняем даты для читаемости
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()


#  График 2: Доля зарегистрированных пользователей
plt.figure(figsize=(12, 5))
plt.plot(
    daily_stats["session_date"],
    daily_stats["registration_share"],
    color="darkorange",
    linewidth=2,
    label="Доля зарегистрированных",
)

plt.title("Динамика доли зарегистрированных пользователей по дням", fontsize=14)
plt.xlabel("Дата сессии", fontsize=12)
plt.ylabel("Доля от общего числа пользователей", fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
````

#### 2.1. Расчёт размера выборки
В рамках курса вы уже рассчитывали размеры выборки и  использовали для этого онлайн-калькулятор. В этом задании предлагаем воспользоваться готовым кодом и рассчитать необходимое для вашего эксперимента количество пользователей.

Для этого установите в коде ниже следующие параметры:

- Уровень значимости — 0.05.

- Вероятность ошибки второго рода — 0.2.

- Мощность теста.

- Минимальный детектируемый эффект, или MDE, — 3%. Обратите внимание, что здесь нужно указать десятичную дробь, а не процент.

При расчёте размера выборки используйте метод `solve_power()` из класса `power.NormalIndPower` модуля `statsmodels.stats`.

Запустите ячейку и изучите полученное значение.

````python
# Задайте параметры:
alpha = 0.05  # Уровень значимости
beta = 0.2  # Ошибка второго рода, часто 1 - мощность
power = 0.8  # Мощность теста
p = 0.3 # Базовый уровень доли
mde = 0.009  # Минимальный детектируемый эффект
effect_size = proportion_effectsize(p, p + mde)

# Инициализируем класс NormalIndPower
power_analysis = NormalIndPower()

# Расчёт размера выборки
sample_size = power_analysis.solve_power(
    effect_size=effect_size,
    power=power,
    alpha=alpha,
    ratio=1
)
````

Здесь важно различать процентные пункты и проценты относительно базового уровня. Если записать mde = 0.03, это будет означать изменение доли, например, с 0.30 до 0.33. - Рассчитайте и выведите на экран процентную разницу в количестве пользователей в группах A и B. Постройте любую удобную визуализацию, на которой будет видно возможное различие двух групп.

Для расчёта процентной разницы воспользуйтесь формулой:
$$P = 100 \cdot  \frac{|A − B|}{A}$$

Считываем датафрейм:

````python
sessions_test_part = pd.read_csv('https://code.s3.yandex.net/datasets/sessions_project_test_part.csv')
sessions_test_part.info()
````

Проверяем распределение пользователей:

````python
#  Выбираем один день для наблюдения (самый первый день в данных)
target_date = sessions_test_part['session_date'].min() 
one_day_data = sessions_test_part[sessions_test_part['session_date'] == target_date]

# Считаем количество уникальных пользователей в каждой группе за этот день
users_by_group = one_day_data.groupby('test_group')['user_id'].nunique()
count_A = users_by_group.get('A', 0)
count_B = users_by_group.get('B', 0)

#  Рассчитываем процентную разницу по заданной формуле
p_difference = 100 * abs(count_A - count_B) / count_A

print(f"Выбранный день для анализа: {target_date}")
print(f"Уникальных пользователей в группе A: {count_A}")
print(f"Уникальных пользователей в группе B: {count_B}")
print(f"Процентная разница: {p_difference:.2f}%")

#  Визуализация различия групп
plt.figure(figsize=(6, 4))
sns.barplot(x=users_by_group.index, y=users_by_group.values)

plt.title(f'Количество уникальных пользователей по группам ({target_date})', fontsize=12)
plt.xlabel('Экспериментальная группа', fontsize=10)
plt.ylabel('Количество уникальных пользователей', fontsize=10)

# Добавляем подписи значений прямо на столбцы
for i, val in enumerate(users_by_group.values):
    plt.text(i, val / 2, f'{val}', ha='center', color='white', fontweight='bold')

plt.show()
````

#### 3.2. Проверка пересечений пользователей
Помимо проверки равенства количества пользователей в группах, полезно убедиться в том, что группы независимы. Для этого нужно убедиться, что никто из пользователей случайно не попал в обе группы одновременно.

- Рассчитайте количество пользователей, которые встречаются одновременно в группах A и B, или убедитесь, что таких нет.

````python
#  Выделяем множества уникальных пользователей для каждой группы
users_A = set(sessions_test_part[sessions_test_part['test_group'] == 'A']['user_id'])
users_B = set(sessions_test_part[sessions_test_part['test_group'] == 'B']['user_id'])

#  Находим их пересечение (пользователи, попавшие в обе группы)
overlapping_users = users_A.intersection(users_B)
overlapping_count = len(overlapping_users)

#  Выводим результат на экран
print(f"Количество пользователей в группе А: {len(users_A)}")
print(f"Количество пользователей в группе B: {len(users_B)}")
print(f"Количество пользователей, попавших в обе группы одновременно: {overlapping_count}")

if overlapping_count == 0:
    print("✅ Отлично! Группы абсолютно независимы, пересечений нет.")
else:
    print(f"⚠️ Внимание! Обнаружено пересечение. {overlapping_count} пользователей видели обе версии.")
````

#### 3.4. Равномерность распределения пользователей по регионам
Теперь убедитесь, что пользователи равномерно распределены по регионам.

Постройте две диаграммы:

- доля каждого региона для пользователей из группы A,

- доля каждого региона для пользователей из группы B.

Постарайтесь добавить на диаграммы все необходимые подписи, пояснения и заголовки, которые позволят сделать вывод о том, совпадает ли распределение регионов в группах A и B. Постарайтесь использовать другой тип диаграммы, не тот, что в прошлом задании.

````python
#  Считаем количество уникальных пользователей по регионам для каждой группы
region_A = (
    sessions_test_part[sessions_test_part['test_group'] == 'A']
    .groupby('region')['user_id']
    .nunique()
)
region_B = (
    sessions_test_part[sessions_test_part['test_group'] == 'B']
    .groupby('region')['user_id']
    .nunique()
)

#  Переводим абсолютные значения в проценты (доли) для корректного сравнения
pct_A = (region_A / region_A.sum() * 100).rename('Группа A')
pct_B = (region_B / region_B.sum() * 100).rename('Группа B')

#  Объединяем данные в одну таблицу для удобства построения графика через seaborn
df_regions = pd.concat([pct_A, pct_B], axis=1).reset_index()
df_melted = df_regions.melt(
    id_vars='region',
    value_vars=['Группа A', 'Группа B'],
    var_name='Группа',
    value_name='Доля пользователей, %',
)

# Сортируем регионы по убыванию популярности в группе А для красивого отображения
order = pct_A.sort_values(ascending=False).index

#  Строим визуализацию
plt.figure(figsize=(12, 6))
sns.barplot(
    data=df_melted,
    y='region',
    x='Доля пользователей, %',
    hue='Группа',
    order=order,
    palette=['#4f81bd', '#c0504d'],
)

# Настройка подписей и заголовков
plt.title(
    'Сравнение распределения пользователей по регионам',
    fontsize=16,
    fontweight='bold',
    pad=20,
)
plt.xlabel('Доля от общего числа пользователей в группе (%)', fontsize=12)
plt.ylabel('Регион', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.legend(title='Экспериментальные группы', fontsize=11)

plt.tight_layout()
plt.show()
````

Длина синих (Группа A) и красных (Группа B) полос для каждого региона визуально одинакова. Незначительные отклонения в доли процента укладываются в рамки естественной случайной погрешности.

Барьерные метрики:
 - "Конверсия в регистрацию"  - в контексте данного эксперимента оценивает долю пользователей, которые успешно зарегистрировались в приложении, от общего числа пользователей, совершивших свою первую сессию. В имеющемся датасете эта метрика представлена бинарной колонкой registration_flag (1 — зарегистрировался, 0 — нет).Поскольку наш новый алгоритм вовлекает новичков в просмотр контента в их самую первую сессию (good_session), регистрация выступает в роли главного «предохранителя» бизнеса. Задача алгоритма рекомендаций — развлечь пользователя, но главная задача бизнеса — превратить анонимного посетителя в зарегистрированного клиента. 
 - Доля «одноразовых» пользователей. Процент людей, у которых session_number == 1 осталась единственной сессией, и они больше никогда не открыли приложение.

#### 4.3. Сравнение доли успешных сессий

Перейдем к анализу ключевой метрики — доле успешных первых сессий.

Используйте созданный на первом шаге задания столбец `good_session` и рассчитайте долю успешных первых сессий для выборок A и B, а также разницу в этом показателе. Полученный вывод отобразите на экране.

````python
#  Фильтруем датафрейм, оставляя только первые сессии пользователей
first_sessions = sessions_test[sessions_test['session_number'] == 1]

#  Рассчитываем долю успешных сессий (конверсию) для групп A и B среди первых сессий
conversion_by_group = first_sessions.groupby('test_group')['good_session'].mean()

conv_A = conversion_by_group.get('A', 0)
conv_B = conversion_by_group.get('B', 0)

#  Рассчитываем абсолютную и относительную разницу между тестом и контролем
absolute_diff = conv_B - conv_A
relative_diff = (absolute_diff / conv_A) * 100

#  Выводим результаты на экран
print(f"Доля успешных первых сессий в группе A (Контроль): {conv_A * 100:.2f}%")
print(f"Доля успешных первых сессий в группе B (Тест): {conv_B * 100:.2f}%")
print(f"Абсолютная разница в конверсии: {absolute_diff * 100:+.2f} п.п.")
print(f"Относительное изменение (прирост) в группе B: {relative_diff:+.2f}%")
````

Если доля успешных сессий в одной из групп выше, это может говорить о том, что изменения, протестированные в этой группе, положительно повлияли на пользовательский опыт. А значит, они могут быть внедрены в основную версию продукта.

Без такой проверки невозможно оценить реальное влияние нововведения, даже если распределение сессий по группам было корректным.

#### 4.4. Насколько статистически значимо изменение ключевой метрики

На предыдущем шаге вы убедились, что доля успешных первых сессий для выборок A и B ухудшилась на 0.1%, но делать выводы только на основе этого значения будет некорректно. Для принятия решения всегда необходимо отвечать на вопрос: является ли это изменение статистически значимым.

- Используя статистический тест, рассчитайте, является ли изменение в метрике доли успешных сессий статистически значимым.

- Выведите на экран полученное значение p-value и свои выводы о статистической значимости. Напомним, что уровень значимости в эксперименте был выбран на уровне 0.05.

````python
#  Отфильтруем данные по первой сессии
first_sessions = sessions_test[sessions_test['session_number'] == 1]

#  Посчитаем количество успешных сессий в каждой группе
successes_A = first_sessions[first_sessions['test_group'] == 'A']['good_session'].sum()
successes_B = first_sessions[first_sessions['test_group'] == 'B']['good_session'].sum()

#  Посчитаем общее количество первых сессий (наблюдений) в каждой группе
trials_A = first_sessions[first_sessions['test_group'] == 'A']['user_id'].count()
trials_B = first_sessions[first_sessions['test_group'] == 'B']['user_id'].count()

#  Проводим Z-тест и сохраняем p-value
z_stat, p_value = proportions_ztest([successes_B, successes_A], [trials_B, trials_A], alternative='larger')

# Выводим  p-value
print(f"p-value равен {p_value:.4f}")
````

Для односторонней проверки гипотезы можно сформулировать так:

* H₀: доля успешных сессий в группе B не выше, чем в группе A.
* H₁: доля успешных сессий в группе B выше, чем в группе A.

При таком варианте проверки полученное значение p-value = 0.578, что больше уровня значимости 0.05. Значит, статистически значимого роста доли успешных сессий в группе B не обнаружено.

Для двусторонней проверки гипотезы можно сформулировать так:

* H₀: доля успешных сессий в группе B не отличается от доли успешных сессий в группе A.
* H₁: доля успешных сессий в группе B отличается от доли успешных сессий в группе A.

При таком варианте проверки полученное значение p-value = 0.8433, что также больше 0.05. Значит, статистически значимых различий между группами нет.

Важно не интерпретировать длительность теста как гарантию обнаружения любого эффекта. Мощность теста рассчитывается для заранее заданного ожидаемого эффекта, а фактическая разница между группами оказалась очень маленькой и отрицательной.

В обоих вариантах интерпретация остаётся одинаковой: новый алгоритм не показал ожидаемого улучшения, поэтому текущую версию изменений нельзя считать успешной.
